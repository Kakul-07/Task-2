### Filling Missing Values

```python
categorical_missing = [
    "loan_limit",
    "approv_in_adv",
    "loan_purpose",
    "Neg_ammortization",
    "age",
    "submission_of_application"
]
for col in categorical_missing:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
numeric_missing = [
    "rate_of_interest",
    "Interest_rate_spread",
    "Upfront_charges",
    "term",
    "property_value",
    "income",
    "LTV",
    "dtir1"
]
for col in numeric_missing:
    df[col] = df[col].fillna(df[col].median())
```

### Outliers Handling

```python
outlier_cols = [
    "loan_amount",
    "rate_of_interest",
    "Interest_rate_spread",
    "Upfront_charges",
    "property_value",
    "income",
    "LTV",
    "dtir1"
]
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower=lower, upper=upper)
```
